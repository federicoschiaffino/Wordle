```python
def evaluate_guess(word_try, word_guess):
    # Initialise results: 0 = grey, 1 = yellow, 2 = green
    res = [0, 0, 0, 0, 0]

    # Count the occurrences of each letter in both words
    try_count = Counter(word_try)
    guess_count = Counter(word_guess)

    # Check whether each letter is absent, in the wrong position, or correct
    for idx in range(len(word_try)):
        if word_try[idx] not in word_guess:
            res[idx] = 0
            
        if word_try[idx] in word_guess and word_try[idx] != word_guess[idx]:
            res[idx] = 1
            
        if word_try[idx] == word_guess[idx]:
            res[idx] = 2

    # Correct the results for words containing repeated letters
    for letter in try_count.keys():
        if try_count[letter] == 2 and guess_count[letter] == 1:
            dup_idx = [index for index, value in enumerate(word_try) if value == letter]

            if res[dup_idx[0]] == 2 and res[dup_idx[-1]] == 1:
                res[dup_idx[-1]] = 0
            if res[dup_idx[0]] == 1 and res[dup_idx[-1]] == 2:
                res[dup_idx[0]] = 0
            if res[dup_idx[0]] == 1 and res[dup_idx[-1]] == 1:
                res[dup_idx[-1]] = 0

        # Handle cases where the guessed word contains three copies of a letter
        if try_count[letter] == 3 and guess_count[letter] == 1:
            dup_idx = [index for index, value in enumerate(word_try) if value == letter]
            outcome = [res[i] for i in dup_idx]
            
            if 2 in outcome:
                for i in dup_idx:
                    if res[i] == 1:
                        res[i] = 0

        # Handle cases where three guessed letters match two occurrences in the target
        if try_count[letter] == 3 and guess_count[letter] == 2:
            dup_idx = [index for index, value in enumerate(word_try) if value == letter]
            outcome = [res[i] for i in dup_idx]

            if Counter(outcome)[2] == 0:
                res[dup_idx[-1]] = 0
            
            if Counter(outcome)[2] == 1:
                if res[dup_idx[-1]] == 1:
                    res[dup_idx[-1]] = 0
                else:
                    res[dup_idx[-2]] = 0
                    
            if Counter(outcome)[2] == 2:
                for i in dup_idx:
                    if res[i] == 1:
                        res[i] = 0

    # Return the Wordle result as a tuple
    return tuple(res)
```

```python
def calculate_entropy(word_try, word_list): 
    # Calculates the entropy of each possible guess based on all possible Wordle outcomes
    word_dict = {}
    out_dict = {}
    entropy_dict = {}
    prob_dict = {}

    # Calculate the Wordle outcome for each possible guess and target word
    for word in word_try:
        for w in word_list:
            out = evaluate_guess(word, w)

            # Group possible target words by their Wordle outcome
            if word not in out_dict.keys():
                out_dict[word] = {out: [w]}
            else:
                if out not in out_dict[word].keys():
                    out_dict[word][out] = [w]
                else:
                    out_dict[word][out].append(w)
        
        # Calculate the probability and entropy of each possible outcome
        for p in out_dict[word].keys():
            if word not in prob_dict.keys():
                prob_dict[word] = {p: (len(out_dict[word][p]) / len(word_list))}
                entropy_dict[word] = {
                    p: (len(out_dict[word][p]) / len(word_list)) 
                    * math.log2(len(word_list) / len(out_dict[word][p]))
                }
            else:
                prob_dict[word][p] = len(out_dict[word][p]) / len(word_list)
                entropy_dict[word][p] = (
                    len(out_dict[word][p]) / len(word_list)
                ) * math.log2(len(word_list) / len(out_dict[word][p]))
                
            # Sum the entropy of all possible outcomes for each word
            word_dict[word] = sum(entropy_dict[word].values())

    # Return the total entropy, outcomes, probabilities and individual entropies
    return word_dict, out_dict, prob_dict, entropy_dict
```

```python
from tabulate import tabulate
from termcolor import colored

def interactive_wordle_solver(word_list, guess_list, entropy_df=None, out_dict=None, prob_dict=None, entropy_dict=None, i=0):
    i += 1

    print("=" * 50)
    print(colored("🎲 Welcome to Wordle Solver 🎲", "cyan", attrs=["bold"]))
    print("=" * 50)

    # Calculate entropy for all valid guesses against the remaining possible answers
    word_dict, out_dict, prob_dict, entropy_dict = calculate_entropy(guess_list, word_list)

    ent_df = pd.DataFrame(word_dict.items(), columns=['word', 'entropy'])
    ent_df['answer'] = [1 if x in word_list else 0 for x in ent_df['word']]

    if i == 1:
        print(colored("✨ Suggested Best Words to Start With ✨", "yellow"))
    else:
        print(colored("✨ Suggested Words Based on Current Guesses ✨", "yellow"))

    best_words = ent_df.sort_values(['entropy', 'answer'], ascending=[False, False]).head()
    print(tabulate(best_words, headers=["Word", "Entropy", "Possible Answer"], tablefmt="fancy_grid"))

    print()
    new_word = input(colored("💡 Please, enter your next word guess: ", "green"))
    new_wordle_outcome = input(colored("🔍 Enter the Wordle outcome (2 = green, 1 = yellow, 0 = grey): ", "green"))

    # Convert the Wordle result into a tuple and handle invalid input
    try:
        wordle_outcome = tuple(int(x) for x in new_wordle_outcome)
    except ValueError:
        print(colored("⚠️ Invalid input. Please enter a sequence of 0s, 1s, and 2s.", "red"))
        return interactive_wordle_solver(word_list, guess_list, i=i - 1)

    # Check that the guess exists in the guess list
    if new_word not in guess_list:
        print(colored("⚠️ That word is not in the guess list.", "red"))
        return interactive_wordle_solver(word_list, guess_list, i=i - 1)

    # Filter the possible answers based on the latest outcome
    new_word_list = out_dict[new_word][wordle_outcome]

    print()
    print(colored(f"🔑 Probability of this outcome: {prob_dict[new_word][wordle_outcome]:.4f}", "blue"))
    print(colored(f"📊 Entropy for this guess: {entropy_dict[new_word][wordle_outcome]:.4f}", "blue"))
    print(colored(f"There are {len(new_word_list)} possible matches.", "blue"))

    # Return the answer if only one possible word remains
    if len(new_word_list) == 1:
        print(colored("🎉 The word is:", "green", attrs=["bold"]))
        return new_word_list[0]

    # Display the remaining possibilities if there are 15 or fewer
    if len(new_word_list) <= 15:
        print(colored("📋 Possible Matches:", "yellow"))
        print(new_word_list)

    # Continue with the reduced list of possible answers
    return interactive_wordle_solver(new_word_list, guess_list, i=i)
```

```python
def wordle(tries: list, yellow_letter, green_letter):
    wrong = []
    for w in word_set:
        for word in tries:
            wrong.append(word)
            for letter in word:
                if letter not in yellow_letter[tries.index(word)] and letter not in green_letter[tries.index(word)]:
                    if letter in w:
                        wrong.append(w)
                if letter in yellow_letter[tries.index(word)]:
                    if w[word.index(letter)] == letter:
                        wrong.append(w)
                    if letter not in w:
                        wrong.append(w)
                if letter in green_letter[tries.index(word)]:
                    if w[word.index(letter)] != letter:
                        wrong.append(w)
                        
    
    res = [x for x in word_set if x not in wrong]
    return res
                    
                    
```
